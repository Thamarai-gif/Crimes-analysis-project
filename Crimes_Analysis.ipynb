```python
numeric_cols = df.select_dtypes(include=['int64']).columns
skew_values = {}
for col in numeric_cols:
    skew_values[col] = df[col].skew()

print("Skewness values:", skew_values)

max_skew_col = max(skew_values, key=lambda k: abs(skew_values[k]))
print("Column with highest absolute skewness:", max_skew_col)
```

```python
col = 'Year'
Q1 = df[col].quantile(0.25)
Q3 = df[col].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print(f"{col} -> Q1: {Q1}, Q3: {Q3}, IQR: {IQR}")
print(f"Lower bound: {lower_bound}, Upper bound: {upper_bound}")

outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
print("Number of outliers:", outliers.shape[0])
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

numeric_df = df.select_dtypes(include=['number'])
corr_matrix = numeric_df.corr()
plt.figure(figsize=(6,4))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")

plt.title("Correlation Matrix of Numeric Columns", fontsize=14)
plt.show()
```

```python
import pandas as pd

numeric_cols = df.select_dtypes(include=['number']).columns
skewness = df[numeric_cols].skew()

top_skewed = skewness.abs().sort_values(ascending=False).head(2).index
for col in top_skewed:
    mean_val = df[col].mean()
    median_val = df[col].median()
    print(f"{col}: mean = {mean_val:.2f}, median = {median_val:.2f}")

for col in top_skewed:
    df[col] = df[col].fillna(df[col].median())

print(df[top_skewed].isnull().sum())
    
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

numeric_df = df.select_dtypes(include=['number'])

pearson_corr = numeric_df.corr(method='pearson')

spearman_corr = numeric_df.corr(method='spearman')
diff = (spearman_corr - pearson_corr).abs()

print("Pearson Correlation Matrix:\n", pearson_corr, "\n")
print("Spearman Correlation Matrix:\n", spearman_corr, "\n")
print("Absolute Difference Table:\n", diff, "\n")


plt.figure(figsize=(6,4))
sns.heatmap(spearman_corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Spearman Rank Correlation Matrix", fontsize=14)
plt.tight_layout()
plt.show()
```
